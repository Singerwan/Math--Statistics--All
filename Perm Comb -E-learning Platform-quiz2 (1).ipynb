# Rowa E-learning Platform-quiz1

```python
# Nine out  12 members of the school's Geography club are to taken out on a study tuor.
# given that there are 7 boyds and 5 girls and at least 3 girls have to go for the tour 
# 1  find the number of ways in which the selection of students can be done 
# 2  if there are 2 sisters in the club, who are definitely selected to go , 
# --------------in how many ways can the remaining students to selected?

#############Question 1 #############################################################
ntotal =12
nboys  =7
ngils  = 5

rtotal      =9
rgirls_list=np.arange(3,ngils+1)
rboys_list =rtotal-rgirls_list

ques1_list=[]
ques1_ans=0
ques1_latex=[]
for i in range(len(rboys_list)):
    ques1_ans=ques1_ans+ math.comb(nboys,rboys_list[i]) * math.comb(ngils,rgirls_list[i])  
    ques1_list.append(math.comb(nboys,rboys_list[i]) * math.comb(ngils,rgirls_list[i]))
    ques1_latex.append("$_{}C_{} \\ \\cdot \\  _{}C_{}$".format(nboys,rboys_list[i],ngils,rgirls_list[i]))
    
#############Question 2 #############################################################
incl=2

ngils_2_incld   = ngils-incl
rgirls_list2    = np.arange(1,ngils_2_incld+1)
rboys_list2     = nboys-rgirls_list2

ques1_ans2=0
ques2_list=[]
ques2_latex=[]
for i in range(len(rgirls_list2)):
    ques1_ans2 = ques1_ans2  +  math.comb(nboys,rboys_list2[i]) * math.comb(ngils_2_incld,rgirls_list2[i])  
    ques2_list.append( math.comb(nboys,rboys_list2[i]) * math.comb(ngils_2_incld,rgirls_list2[i]))
    ques2_latex.append("$_{}C_{} \\ \\cdot \\  _{}C_{}$".format(nboys,rboys_list2[i],ngils_2_incld,rgirls_list2[i]))
ques1_ans2

#############answer 1 & 2 #############################################################
print("ans1 :{}ways\nans2 :{}ways".format(ques1_ans,  ques1_ans2))


#############df creation #############################################################
df= pd.DataFrame(data={ "$n_{\\Sigma}$":ntotal,
                        "$n_{boys}$"   :nboys,
                        "$n_{girls}$"  :ngils,
                        "$r_{\\Sigma}$":rtotal ,
                        "$r_{1-grils}$":rgirls_list,
                        "$r_{1-boys}$":rboys_list,
                        "$latex1$":ques1_latex,
                        "$ans1$":ques1_list,
                        "$ans1\\Sigma$":ques1_ans,
                        "|":"|",
                        "g2-incl":2,
                        "$r2_{gilrs}$":rgirls_list2,
                        "$r2_{boys}$":rboys_list2,
                        "$latex2$":ques2_latex,
                        "$ans2$":ques2_list,
                         "$ans2\\Sigma$":ques1_ans2},
                        index=np.arange(1,4))
df.index.name="2Qs"
df.columns.name= "NO"

# 
df.columns

array=[['n', 'n', 'n', 'r1',
       'r1', 'r1', 'q1', 'q1', 'q1',
       '|', 'r2', 'r2', 'r2', 'q2', 'q2',
       'q2'],
         ['$n_{\\Sigma}$', '$n_{boys}$', '$n_{girls}$', '$r_{\\Sigma}$',
               '$r_{1-grils}$', '$r_{1-boys}$', '$latex1$', '$ans1$', '$ans1\\Sigma$',
               '|', 'g2-incl', '$r2_{gilrs}$', '$r2_{boys}$', '$latex2$', '$ans2$',
               '$ans2\\Sigma$']]
df.columns=pd.MultiIndex.from_arrays(array)
df
```
