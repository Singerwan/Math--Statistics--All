### Mutually exclusive Events

<hr>
mutually exclusive events
<hr>
A $\cap$  B =0 &rarr; mutually exlusive events :there's no intersection &emsp;<span style="font-size:40px;color:cyan">$\boxed{\circ \\ \circ}$</span> 
<br>P(A or B) = P(A)+P(B)-P(A $\cap$  B)=P(A)+P(B)-0   (double cont  |intersection| needs to be subtracted - )
<br>Example : card withdrawl vs dice roll
<hr>
independent events | flip a coin roll a die|
<hr>
one event happening doesn;t affect the other event's happening<br>
P(A and B) = P(A) $\times$ P(B)
<br>
overlapping - intersection to check whether two events are dependent or not can use the above formula

```python
A1,AnC,C1=sympy.symbols([ "A1","AnC", "C1"])

eq1      =sympy.Eq(0.27,A1+AnC)
eq2      =sympy.Eq(0.2,C1+AnC)
eq3      =sympy.Eq(0.29,A1++AnC+C1)

var_list=[]
ans_list=[]
for var, ans in sympy.solve([eq1,eq2,eq3],[A1,AnC,C1]).items():
    print("{} is: {:.2f}".format(var,ans))
    var_list.append(var)
    ans_list.append(ans)
print("=============ans checked")
#-check the answer 
ans_list.append(0.63)
ans_list.append(0.07)
np.array(ans_list).sum()
```
