### quiz1-python solutions 

```python
row_n = 8
col_n = 4

matrix = np.zeros(( row_n , col_n ),dtype="object")

lst    = [1,2,3,4]
init   = 0
for i in lst:
    init=init+(math.perm(row_n,i)/math.factorial(i))
ans2 = init**col_n  
print("ani    :1 column   {}\nansiii :4 columns  {:,}  key cards".format(int(init),int(ans2)))


matrix=[["","O","","O","","","","O"],
       ["","","","O","","O","O","",],
       ["","","O","","","","",""],
       ["O","","","","O","","",""],]

df= pd.DataFrame(matrix)
dft= df.T
dft.columns=np.arange(1,5)
dft.index  =np.arange(1,9)
dft

def color(x):
    if x =="O":
        return "background-color:red;color:cyan;border:double white 2px;padding:5px;font-size;30px;font-weight:bold"
    elif x=="":
        return "background-color:blue:color:cyan"

dft.style.map(color)
```

### quiz3-python solutions

```python
fact=[2,2,3,4,1]
n   = 12
fact_ini=1
for i in fact:
   fact_ini=fact_ini * math.factorial(i)
ansi = math.perm(n)/fact_ini
ansi
```

```python
objec_n = 6
space_n = 6

group1_lst  =["A","A","D","D","D","D"]
group1_2nd  =["B","B","C","C","C","E"]

facto =list(Counter(group1_lst).values()) + list(Counter(group1_2nd).values())

num   =math.perm(objec_n,space_n)**2
den   =1

for i in facto:
    den=den*math.factorial(i)

ansii =num/den
# reason - it's the product - 1st group and the second group 
# ------------each group is essentially 6 spaces choose 6 houses -  | each style a b c d e doen'st 
# ------------need to be in is consolidated subgroup- they can be mixed with different style subgroup
# ------------thus they are treated as individual items - spaces choose objects 
# ------------yet - the style of each type is identical therefore , each type must be treated as non-distinct 
# ------------items , therefroe, the 6p6 must be divided by the non-distinct item's factorial derived 
# ------------from the occurance acount 
ansii
```

### quiz5-pythhon solutions

```python
x=np.arange(1,9)
y=np.arange(1,3.5,0.5)
s=np.arange(1,41,1).reshape(len(y),len(x))
plt.figure(figsize=(12,6))

rect1= plt.Rectangle((0.69,7.3),1.1,1.2,edgecolor="red",lw=3)
plt.gca().add_patch(rect1)
rect2= plt.Rectangle((0.69,0.5),1.6,1.2,edgecolor="indigo",lw=3)
plt.gca().add_patch(rect2)

plt.ylim(0,9)
plt.xlim(0.5,6)
for j in range(len(x)):
    for i in range(len(y)):
        plt.plot(y[i],x[j],"s",ms=40,mec="blue",mew=2)
        plt.annotate("{}".format(s[i,j]),size=11,xy=(y[i]-0.05,x[j]-0.065))
        if i <2 :
            plt.plot(y[i],x[-1],"o",ms=25,mec="white",mew=2)      
        if  i <3:
            plt.plot(y[i],x[0],"o",ms=25,mec="white",mew=2)     
        plt.annotate("How many possible arrangements \nare there of seating \nMary, Ahmad, Wayne, Elsie and John \nif Mary and Ahmad sit together\nin the front row and the other three\nsit together in one of\nthe other rows?"
                      ,xy=(3.16,0.5),
                      xytext=(3.5,5),size=15,
                      color="blue",bbox=dict(boxstyle="sawtooth,pad=0.5",facecolor="yellow"),
                      arrowprops=dict(arrowstyle="fancy",facecolor="cyan",connectionstyle="arc3,rad=0.01"))
```

```python
ansi  =math.perm(40,5)
ansii =math.perm(2,2)*math.comb(4,3)*7*math.perm(3,3)*math.comb(3,2)
# logic behind the quiz 
# two restrictions are imposed in this quiz, the first is a pair , that's specifically occupying the first row , 
# yet it can shuffle within in side - then the remaining 3 seats in the first row and the pair treated as a single unit 
# is 4p4/3! as the three seats are identical  without ordering mentioned 
# - --- second is the trio - which can be seated at any remaining 7 rows, therefore, 
# ------  2.1 they can also shuffle within themselves, thus 3! , and the remaining two empty seats and this unit is 3p3 / 2! as
# ------- the empty seats area treated as identical , due to the fact no specifiic orders have been mentioned
# =========finnally , because there are 7 duplicated rows, then the ways are multiplied by 7 
ansii
```
