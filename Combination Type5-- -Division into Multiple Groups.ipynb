## Quiz1


<mark>Q1 (W12/qp61/q7(c))<br>
<u>9 different fruit pies</u> are to be divided between 3 people so that each person gets an odd 
number of pies. 
<br>Find the number of ways this can be done.</mark>

<div style="font-size:15px">
We have 9 different fruit pies to divide among 3 people, so that each person receives an odd number of pies.
Step 1: Work out the possible distributions<br>
Since there are 9 pies and each person must receive an odd number, the possible distributions are:
$[(1,1,7),\quad (1,3,5),\quad (3,3,3)]$</div>
<div style="font-size:15px">
We must consider the different arrangements between the 3 people.
<br>
Step 2: Calculate each case
<br>
Case 1: Distribution (1, 1, 7)
 - Choose which person gets 7 pies,
 - then choose the 7 pies<nr>
 - distributing the remaining 2 pies to the other two people :
<br>$[
\frac{^3P_3}{2!}=\frac{3!}{(3-3)2!}=\frac{3!}{2!}=3\times\binom{9}{7}\times\binom{2}{1}\times\binom{1}{1}
][
=3\times36\times2=216
]$
<hr>
Case 2: Distribution (1, 3, 5)

 - Since all three people receive different numbers of pies, 
 - there are \(3!\) ways to assign these numbers.
$[^3P_3=\frac{3!}{(3-3)!)}=\frac{3!}{1!}=3!\times\binom{9}{1}\times\binom{8}{3}\times\binom{5}{5}][=6\times9\times56\times1=3024]$
<hr>
Case 3: Distribution (3, 3, 3)
<br>$[\frac{^3P_3}{(3-3)! \times 3!}=\frac{3!}{3!}=1\binom{9}{3}\times\binom{6}{3}\times\binom{3}{3}][=84\times20\times1=1680]$
<br>
Step 3: Add the cases
$[
216+3024+1680=\boxed{4920}
]$
<hr>
Final answer: 4920 ways
Exam tip:
<br>When dividing different objects among people with restrictions,<br> first list all possible number distributions. 
<br>Then count each distribution separately, taking care not to double-count arrangements.
<hr>
</div>

#### Python Solution 

```python
# 1 1 7
# 3 5 1
# 3 3 3
3 * math.comb(9,7) * math.comb(2,1)*math.comb(1,1) +\
math.perm(3,3) * math.comb(9,3) * math.comb(6,5) * math.comb(1,1)  + \
math.comb(9,3)*math.comb(6,3)*math.comb(3,3) 
```
