## Quiz-27

<div style="color:indigo;background-color:cyan">27.(S22/qp53/q7(b)) <br>
A group of 15 friends visit an adventure park. The group consists of four families.  <br>
<hr>
• Mr and Mrs Kenny and their four children <br> 
• Mr and Mrs Lizo and their three children <br>
• Mrs Martin and her child <br>
• Mr and Mrs Nantes  <br>
<hr>
The group travel to the park in three cars, one containing 6 people, one containing 5 people and <br>
one containing 4 people. The cars are driven by Mr Lizo, Mrs Martin and Mr Nantes respectively. <br>
The group enter the park by walking through a gate one at a time. <br>
(b) In how many different orders can the 15 friends go through the gate if Mr Lizo goes first and 
each family stays together? 

We need to find the number of different orders in which the 15 people enter the gate, given that:
1. Mr Lizo goes first.
2. Each family stays together.
Step 1: Treat each family as one block
There are 4 families:
Family	Number of people
Kenny	6
Lizo	5
Martin	2
Nantes	2

Since each family must stay together, we treat each family as one block.
Step 2: Fix the Lizo family first
Mr Lizo must go first, so the Lizo family must enter first.

The remaining 3 family blocks can be arranged in:
$[
3!=6
]$

ways.
Step 3: Arrange the people within each family
The members of each family can enter in any order.
- Kenny family: \(6!\)
- Lizo family: \(5!\)
- Martin family: \(2!\)
- Nantes family: \(2!\)
However, Mr Lizo must be first within his family, so the Lizo family has only:
$[
4!
]$

possible arrangements.
Step 4: Multiply everything
$[
3!\times6!\times4!\times2!\times2!
]$

$[
=6\times720\times24\times2\times2
]
[
=\boxed{414720}
]$

Final answer: 414,720
Exam tip: When a question says each family stays together, treat each family as a single block first. Then multiply by the number of possible arrangements within each family.
