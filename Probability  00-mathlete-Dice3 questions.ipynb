## Quiz_1

- Two dice are rolled and their sum is recorded, find the probability that the score is
  - a perfect square
  - a prime number
  - the difference between the two scores is 5

### python solution 

#### ans_a

```python
die=np.arange(1,7)
mat=np.zeros((6,6))
for i in range(6):
    for j in range(6):
        mat[i,j]=die[i]+die[j]
count_sqr=0
count_prm=0
mat_flat=mat.flatten()

for i in mat_flat:
    if np.sqrt(i) ==2 or np.sqrt(i) ==3:
        count_sqr=count_sqr+1
    elif i in [2,3,5,7,11]:
        count_prm=count_prm+1
```
