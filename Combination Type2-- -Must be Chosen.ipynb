## Quiz10

The 26 members of the local sports club include Mr. and Mrs. Khan and their son  <br>
Abad. The club is holding a party to celebrate Abad’s birthday, but there is only  <br>
room for 20 people to attend. In how many ways can the 20 people be chosen  <br>

#### Python Solution 

```python
k = 3 
n = 26
r = 20 
math.comb(n-k,r-k)
```
