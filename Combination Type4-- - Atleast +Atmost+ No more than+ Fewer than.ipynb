Q3 (W11/qp63/q4(iii))  <br>
Mary saves her digital images on her computer in three separate folders named <br>‘Family’,

<hr>

<br>

Find the number of different ways in which Mary can choose 6 of these images if there are 
 <br>at
least 3 images from the friends folder and 
 <br>at least 1 image from each of the other two 
folders.




#### Python Solution 

```python
matrx = np.zeros((3,4),dtype="object")

val   = [[3,1,2,6],
         [3,2,1,6],
         [4,1,1,6]]

for i in range(matrx.shape[0]):
    for j in range(matrx.shape[1]):
        matrx[i,j]=val[i][j]
df=pd.DataFrame(matrx,columns=['#Img_fri','#Img_fam','#Img_hol','total_s'],index=np.arange(1,4))

new_cols={ "#fr_folder": np.ones(3)*8,
           "#hol_folder":np.ones(3)*4,
           "#fam_folder":np.ones(3)*3,}
df=df.assign(**new_cols)

latex  =[]
result = 0
for i in range(3):
    result=result + math.comb( int(list(df['#fr_folder'])[i]) ,   int(list(df['#Img_fri'])[i])   ) *\
                    math.comb( int(list(df['#hol_folder'])[i]) ,  int(list(df['#Img_hol'])[i])   ) *\
                    math.comb( int(list(df['#fam_folder'])[i]) ,  int(list(df['#Img_fam'])[i])   ) 

    latex.append("$^{}C_{}\\times^{}C_{}\\times^{}C_{}$".format(   int(list(df['#fr_folder'])[i]),    int(list(df['#Img_fri'])[i]),
                                                                   int(list(df['#hol_folder'])[i]) ,  int(list(df['#Img_hol'])[i]) ,  
                                                                   int(list(df['#fam_folder'])[i]) ,  int(list(df['#Img_fam'])[i])   )   )                                       

df['formula']=latex
df["result"]=result



df[['#fr_folder',"#hol_folder","#fam_folder"]]=df[['#fr_folder',"#hol_folder","#fam_folder"]].map(lambda x: int(x))


mul_col=[    ["r","r","r",'$\\Sigma_s$','n',"n","n","result","result"],
             ['#Img_fri', '#Img_fam', '#Img_hol', 'total_s', '#fr_folder','#hol_folder', '#fam_folder', 'formula', 'result'] ]

df.columns = pd.MultiIndex.from_arrays(mul_col)
df

df.style.set_properties(subset='result',**{"background-color":"red","color":"blue"})
```

```python
math.comb(7,1)*math.comb(6,1)*math.comb(4,2) +\
math.comb(7,1)*math.comb(6,2)*math.comb(4,1) +\
math.comb(7,1)*math.comb(6,3)*math.comb(4,0) +\
math.comb(7,2)*math.comb(6,1)*math.comb(4,1) +\
math.comb(7,2)*math.comb(6,2)*math.comb(4,0) +\
math.comb(7,3)*math.comb(6,1)*math.comb(4,0) 
df =pd.DataFrame(data= {"Ch_n":np.ones(6)*7,"Ch_r":[1,1,1,2,2,3],
                        "Eu_n":np.ones(6)*6,"Eu_r":[1,2,3,1,2,1],
                        "Am_n":np.ones(6)*4,"Am_r":[2,1,0,1,0,0],})
df=df.map(lambda x :int(x))
quiz12val=df.values
result= []
latex = []
for i in range(6):
    result.append(   math.comb( list( df['Ch_n'] )[ i ], list( df['Ch_r'] )[ i ] )*\
                        math.comb( list( df['Eu_n'] )[ i ], list( df['Eu_r'] )[ i ])*\
                        math.comb( list( df['Am_n'] )[ i ], list( df['Am_r'] )[ i ]))
    latex.append( "$^{}C_{} \\times ^{}C_{} \\times ^{}C_{}  $".\
                  format(
                      int( list( df['Ch_n'] )[ i ] ),
                      int( list( df['Ch_r'] )[ i ] ),
                      int( list( df['Eu_n'] )[ i ] ),
                      int( list( df['Eu_r'] )[ i ] ),
                      int( list( df['Am_n'] )[ i ] ),
                      int( list( df['Am_r'] )[ i ] )    )   )
new_col={"result":result,
         "formula":latex,
         "$\\Sigma$":np.array(result).sum()}
df=df.assign(**new_col)
df.style.set_properties(subset="formula", **{"backkground-color":"indigo","color":"cyan","border":"double red 2px","padding":"8px"}).\
         set_properties(subset="result", **{"background-color":"blue","color":"yellow","border":"double white 2px","padding":"8px"}).\
         set_properties(subset="$\\Sigma$", **{"background-color":"red","color":"yellow","border":"double white 2px","padding":"8px"}).\
         highlight_max(subset="Am_r",color="red").highlight_min(subset="Am_r",color="green").\
         highlight_max(subset="Ch_r",color="crimson").highlight_min(subset="Ch_r",color="blue").\
         bar(subset="Eu_r",cmap="flag")
```
