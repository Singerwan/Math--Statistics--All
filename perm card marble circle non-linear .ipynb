```python
import matplotlib.pyplot as plt
import numpy as np 
```

```python

colorlist = []  #color list
for i in range(10000):
    colorlist.append('#%06X' % np.random.randint(0, 0xFFFFFF))  # colors list
```

# Cards 

```python
x=[-1.5,-1,-0.5,0]
y1=[-2.5,-1.5,1.5,2.5]
x1=[-1.5,-1.5,-1.5,-1.5]
plt.figure(figsize=(12,6))
y=[0,0,0,0]
plt.xlim(-2,4)
plt.ylim(-3,3)
markerlist =["$\\clubsuit$","$\\spadesuit$", "$\\diamondsuit$","$\\heartsuit$"]
colorlistc = ["black","black","red","red"]
comments   =["FaceCards 3 *4=12","NumberCards 2-10 9*4=36","A ace 1*4=4","4suits 13*4=52"]
for i in range(len(x)):
    plt.plot(x[i],y[i],marker=markerlist[i],ms=50,mfc=colorlistc[i],mec=colorlistc[i])
    plt.annotate("{}".format(comments[i]),xy=(x1[i],y1[i]),color="blue",
                 size=25,
    bbox=dict(boxstyle="sawtooth,pad=0.5",facecolor=colorlist[i],alpha=0.2))

plt.plot(3.1,-.3,marker="$A$",ms=8,mfc="black",mec="black")
plt.plot(3.65,-2.7,marker="$A$",ms=8,mfc="black",mec="black")

plt.plot(3.1,-.6,marker="$\\diamondsuit$",ms=8,mfc="red",mec="red")
plt.plot(3.66,-2.4,marker="$\\diamondsuit$",ms=8,mfc="red",mec="red")
r=plt.Rectangle((2.85,-2.9),1,3,facecolor="white",edgecolor="black")
plt.plot(3.35,-1.5,marker="$\\diamondsuit$",ms=15,mfc="red",mec="red",mew=1)
plt.gca().add_patch(r)
plt.annotate("Double Count \nAce and Diamond"
             ,xytext=(0.8,0.7),color="blue",
             size=25,
             bbox=dict(boxstyle="circle,pad=0.5",facecolor="cyan",alpha=1),
             xy=( 2.87,-1.5),
             arrowprops=dict(arrowstyle="fancy",facecolor="indigo",connectionstyle="arc3,rad=0.2"))
plt.axis("off")
plt.show()
```

# Marble

```python
# 3 three boxes 

plt.figure(figsize=(12,5))
plt.xlim(0,6)
plt.ylim(-1,3)

size=15
ms  =140
h_m =1
h_l =0.5
h_u =1.5
# -first bag contains 2 red 3 black 1 white
plt.plot(1,1.1,marker="s",mfc="white",mec="indigo",ms=ms,mew=3)
plt.plot(1,h_m,marker="o",mfc="white",mec="black",ms=size)
plt.plot([0.5,1.5],[h_l,h_l],marker="o",mfc="red",mec="red",ls="",ms=size)
plt.plot([0.5,1,1.5],[h_u,h_u,h_u],marker="o",mfc="black",mec="black",ls="",ms=size)
# -second bag contains 3 red 1 black 2 white
plt.plot(3,1.1,marker="s",mfc="white",mec="green",ms=ms,mew=3) # mew marker edge width 
plt.plot(3,h_u,marker="o",mfc="black",mec="black",ms=size)
plt.plot([2.5,3.5],[h_m,h_m],marker="o",mfc="white",mec="black",ls="",ms=size)
plt.plot([2.5,3,3.5],[h_l,h_l,h_l],marker="o",mfc="red",mec="red",ls="",ms=size)

# -third bag contains 1 red 2 black 3 white
plt.plot(5,1.1,marker="s",mfc="white",mec="magenta",ms=ms,mew=3)
plt.plot([4.5,5,5.5],[h_l,h_l,h_l],marker="o",mfc="black",mec="black",ls="",ms=size)
plt.plot([4.5,5,5.5],[h_m,h_m,h_m],marker="o",mfc="white",mec="black",ls="",ms=size)
plt.plot(5,h_u,marker="o",mfc="red",mec="red",ls="",ms=size)

plt.axis("off")
plt.show()
```
