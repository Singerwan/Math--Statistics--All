# Mathlete Probability

## Quiz_3

- A biased die has faces numbered 1 to 6. The probabilities of the die landing on 1,3, or 5 are each equal to 0.1.
- The probability of the die landing on 2 or 4 are each equal to 0.2, The die is thrown twice.
- Find the probability that the sum of the number it lands on is 9.
- $\frac{1}{10} \ \times \frac{2}{10} +\frac{2}{10} \ \times \frac{1}{10}\quad  +\quad   \frac{1}{10} \times \frac{3}{10}  +  \frac{3}{10} \ \times \frac{1}{10} $
- $\quad\quad54  \quad \quad \quad 54 \quad \quad \quad \quad \quad \quad \quad  36 \quad \quad\quad 63$

### python solution 

```python
dim = 6
die = np.arange(1,7)
mat = np.zeros((dim,dim),dtype="object")

for i in range(dim):
    for j in range(dim):
        mat[i,j]=die[i]+die[j]
df=pd.DataFrame(data    =mat,
                index   =list(map(lambda x: "Die2_Side{}".format(x), np.arange(1,7) )),
                columns =list(map(lambda y :"Die1_Side{}".format(y), np.arange(1,7))))
df.style.map(lambda x : "color:cyan;border:double white 2px;text-align:center" if x==9 else "color:white")
```
