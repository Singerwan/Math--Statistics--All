```python
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
markerlist=[".",",","o","v","^","<",">","1","2","3","4","8","s","p","P","*","h","H","+","x","X","D","d","|","_","1","2","3","4","8","none","$\\sigma$"]

hatchlist=['//', '\\\\', '||', '--', '++', 'xx', 'oo', 'OO', '..', '**','/o', '\\|', '|*', '-\\', '+o', 'x*', 'o-', 'O|', 'O.', '*-', '/', '\\', '|', '-', '+', 'x', 'o', 'O', '.', '*']   #hatch list plt matplotlib only 

colorlist = []  #color list
for i in range(10000):
    colorlist.append('#%06X' % np.random.randint(0, 0xFFFFFF))  # colors list

print("color : {} marker : {} hatch : {}".format(colorlist[0],markerlist[0],hatchlist[0]))
```

```python
lw1=6
np.random.seed(0)
data=np.random.randn(1000000)
plt.figure(figsize=(12,6))
sns.kdeplot(data)
sample_f=10000
sample_n=30
means   =[]
for i in range(sample_f):
    means.append(np.mean(np.random.choice(data,sample_n,replace=False)))
sns.kdeplot(means)
sample_d=np.random.choice(data,sample_n,replace=False)

plt.plot(np.mean(sample_d)-0.15,1,marker="*",ms=20,mfc="cyan",mec="indigo")
plt.plot(np.std(sample_d)-0.15,1,marker="*",ms=20,mfc="green",mec="red",lw=0.1)

for i in range(sample_n):
    plt.plot(    sample_d[i], np.ones(sample_n)[i],
                 marker=markerlist[i],mfc=colorlist[i],mec="black" ,alpha=0.4)
plt.axvline(0,0,0.9,ls="--",lw=10,alpha=0.3,color="grey")
plt.axvline(0,0,0.94,ls="-",lw=1,alpha=1,color="indigo")


# ======================s=============================================================
plt.plot([np.mean(sample_d)-0.15,np.std(sample_d)-0.15],[1.07,1.07],
         lw=lw1,
         color="cyan",label="Sample STD : s")
plt.annotate(  "Sample STD : s",
                xy=(np.mean(sample_d)-0.15,1.07),
                xytext=(-5,1.8),size=20,color="indigo",
                bbox=dict(boxstyle="sawtooth,pad=0.5",facecolor="cyan",edgecolor="cyan"),
                arrowprops=dict(arrowstyle="fancy",facecolor="cyan",edgecolor="cyan",connectionstyle="arc3,rad=0.1")
            )

# =========================SEM=============================================================
plt.plot([np.mean(sample_d)-0.15,0],[0.92,0.92],
          lw=lw1,
         color="green",label="Standard Error of the sample means:SEM")
plt.annotate(  "Standard Error of \nthe sample means:SEM",
                xy=(0,0.92),
                xytext=(1.5,0.3),size=15,
                bbox=dict(boxstyle="sawtooth,pad=0.5",facecolor="green",edgecolor="green"),
                arrowprops=dict(arrowstyle="fancy",facecolor="green",edgecolor="green",connectionstyle="arc3,rad=0.1")
            )

# ========================Sigma =============================================================
plt.plot([-1,0],[0.23,0.23],lw=lw1,color="red",label="Population STD : $\\sigma$")
plt.annotate(  "Population STD : $\\sigma$",
                xy=(-1,0.23),
                xytext=(-5,0.3),size=20,
                bbox=dict(boxstyle="sawtooth,pad=0.5",facecolor="red",edgecolor="red"),
                arrowprops=dict(arrowstyle="fancy",facecolor="red",edgecolor="red",connectionstyle="arc3,rad=0.1")
            )
plt.legend()
plt.title("STD VS SEM",size=30)
plt.annotate("sample data with a fixed \n size n is graphically \nexpressed as markers",
             xy=(1.2,1.25),size=15,color="white",
             bbox=dict(boxstyle="DArrow",facecolor="teal",edgecolor="brown"))
plt.savefig("s8d11.pdf")
plt.show()
```
