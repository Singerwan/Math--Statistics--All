```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import math
```

```python
# 1
n_rows  =  6
n_cols  =  4
data    =  np.zeros((  n_rows,   n_cols  ))
data 

for i in range(n_cols):
    for j in range(n_rows):
        data[j,i]=np.arange(1,n_rows+1)[j]

df=pd.DataFrame(data,index=np.arange(1,7,1),
                columns=["First_Failure","Second_Failure","Third_Failure","Fourth_Failure"])
df.loc["Probaility",:]=5/6
df.loc["Probaility","Fourth_Failure"]=1/6
df["Prob"]=df.product(axis=1)
df=df.round({ "First_Failure" :2,
              "Second_Failure":2,
              "Third_Failure" :2,
              "Fourth_Failure":2,           
              "Prob":5}             )
df
df1=df.iloc[0:6,0:4].map(lambda x: str(int(x)))
df2=pd.DataFrame(df.iloc[6,0:4].map(lambda x:str(x)))
df3=pd.DataFrame( df.loc[:,"Prob"].map( lambda x:str(x) ) )
df2=df2.T
df_all1=pd.concat((df1,df2),axis=0)
df_all2=pd.concat((df_all1,df3),axis=1)
df_all2.loc[:6,"Prob"]=["prob of 1st 6 on the 4th roll a 6-sided die?",
                        "$p(X=x)=q^{X-1} \\cdot p $",
                        "|Geometric Distribution| x = 4",
                        "p = 1 '6' out of six-sided $\\frac{1}{6}$ q=1-p ",
                        "multiple 2dice 6*6=36 matrix 4 rolls- ",
                        "nd1vector 2matrix 3cluster 4groups"]
df_all2
```

```python
# 5
matrix_n = 6
data = np.zeros((matrix_n,matrix_n))
for i in range(data.shape[0]):
    for j in range(data.shape[1]):
        data[i,j]=(i+1)+(j+1)
df= pd.DataFrame(data,index=np.arange(1,7),columns=np.arange(1,7))
df1=df.map(lambda x: str(int(x)))


def styl(x):
    if x=="4":
        return "background-color:cyan;color:indigo"
    elif x=="$\\frac{3}{36}$":
        return "background-color:yellow;color:black;font-size:20px;font-weight:bold"
    elif x=="2":
        return "background-color:blue;color:cyan"
    elif x=="12":
        return "background-color:green;color:yellow"
    else:
        return "color:red"

unqiue = np.unique(df1.values)

count = 0 
for i in range(6):
    for j in range(6):
        if df1.values[i,j] =="4":
                count =     count +1 
count
posou=data.shape[0]*data.shape[1]

################### prob of x<8 ====================
cal_rep=np.arange(1,8)
p=1/12
q=1-p
power_cal=cal_rep-1
result_cal=[]

for i in range(len(cal_rep)):
    result_cal.append(q**cal_rep[i]*p)
    
probLT8=np.array(result_cal).sum()  # 1-(11/12)**7 simple solution 

print("{} over {}".format(count,posou))
df1["Prob"]=["mean:12 $\\frac{1}{\\frac{1}{12}}$",
             "Prob of 4 on 6th trhow : $(\\frac{11}{12})^{(6-1)} \\cdot \\frac{1}{12}$ ",
             "Prob calculated using the above formula is :  {:.6f}%".format( ((11/12)**(6-1) ) * (1/12) *100 ) ,
             "Prob x$\\lt 8$  : {:0.2f}".format(probLT8),
             "","$\\frac{3}{36}$"]
df1.index.name="Die1 6-sided fair die"
df1.columns.name="Die2 6-sided fair die"
df1.style.map(styl)
```
