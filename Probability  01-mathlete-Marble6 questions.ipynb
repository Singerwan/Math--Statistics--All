# Mathlete Probability

## Quiz_1

- A bag contains 5 red,2 blue and 3 green balls.Two balls are chosen at random without replacement.Find the probability that 
  - the one ball is blue
  - both balls are red
  - first is red and second is green
  - one is red the other is green
  - both are of the same color
  - both are of different colors

### python solution 

```python
from imojify import imojify
from matplotlib import pyplot as plt 
from matplotlib.offsetbox import OffsetImage,AnnotationBbox

dfig,ax=plt.subplots(1,1,figsize=(18,9))
ax.plot(0,0,"")
ax.set_xlim(0,11)
ax.set_ylim(0,3)

emjisss  =["🔴🔴🔴🔴🔴","🔵🔵","🟢🟢🟢"]
x=[1,3,4.5,5,3,5,7]
y=[2.5,2.5,2.5,1,2,2,2]
for i in range(len(emjisss)):
    img = plt.imread(imojify.get_img_path(emjisss[i]))
    ax.add_artist( AnnotationBbox(
                                   OffsetImage(
                                                img, zoom=0.08),
                   (x[i]+0.5, y[i]+0.15),  frameon=True, pad=0.5))
ax.plot(0.5,1,marker="o",ms=50)
ax.plot([0.5,2,0.5,2,0.5,2],
        [1,1,1,1.5,1,0.5],ls="--",color="indigo",lw=4)        
colors_0_list=["red","blue","green"]
colors=["red  \\frac{5}{10}","blue \\frac{2}{10}","green \\frac{3}{10}"]
x1=[2,2,2]
y1=[1.5,1,0.5]
for i in range(3):
    ax.plot(x1[i]+0.5,y1[i],marker="${}$".format(colors[i]),ms=(i+5)*10,mfc=colors_0_list[i],mew=0.02)
    ax.plot(x1[i]+1,y1[i],marker="o",ms=20)
ax.plot([3,4.5,3,4.5,3,4.5],
        [1.5,1.7,1.5,1.5,1.5,1.3],ls="-.",color="indigo",lw=2)        
ax.plot([3,4.5,3,4.5,3,4.5],
        [1,1.2,1,1,1,0.8],ls="-.",color="indigo",lw=2)          
ax.plot([3,4.5,3,4.5,3,4.5],
        [.5,.7,.5,.5,.5,.3],ls="-.",color="indigo",lw=2)        

colors1=[  "red  \\frac{4}{9}","blue \\frac{2}{10}","green \\frac{3}{10}",
           "red  \\frac{5}{10}","blue \\frac{1}{9}","green \\frac{3}{10}",
           "red  \\frac{5}{10}","blue \\frac{2}{10}","green \\frac{2}{9}",]
x2=np.ones(9)*4.5
y2=[1.7,1.5,1.3,1.2,1,0.8,0.7,0.5,0.3]
colors_1_list=["red","blue","green","red","blue","green","red","blue","green"]
for j in range(len(colors1)):
    plt.plot(x2[j]+0.2,y2[j],marker="${}$".format( colors1[j]),ms=35,mec="black",mew=0.03,mfc=colors_1_list[j])

# plt.axis("off")
ax.set_title("Tree Diagram",color="red",size=60,y=0.9,x=0.75)
ax.plot(9.8,1.1,marker="*",ms=110,mfc="red",mec="blue",mew=5)
ax.plot(9.8,1.1,marker="$d$",ms=20,mfc="white",mec="indigo",mew=1)
ax.annotate("""A bag contains 5 red,2 blue and 3 green balls.
Two balls are chosen at random without replacement.
Find the probability that
a. the one ball is blue  :ans-a $\\frac{2}{10}$
b. both balls are red    :ans-b $\\frac{5}{10} \\ \\times  \\ \\frac{5-1^{one \\ taken}}{10-1^{one \\ taken}}(\\frac{4}{9})$
c. first is red and second is green :ans-c $\\frac{5}{10} \\times \\ \\frac{3}{9}$
d. one is red the other is green 
   :difference between c and d is that
   d has two scenarios 1 rg 2 gr
   ans-d $\\frac{5}{10} \\times \\frac{3}{9}$ + $\\frac{3}{10} \\times \\frac{5}{9}$
e. both are of the same color 
   ans-e 2r+2b+2g  $\\frac{5}{10} \\times \\frac{4}{9}+ \\frac{2}{10} \\times \\frac{1}{9} +\\frac{3}{10} \\times \\frac{2}{9}$
f. both are of different colors 
   ans-f 1-both2 balls are of the same color
   1- quiz-subquuize-e""",size=19,
            xy=(5.2,0.2),bbox=dict(boxstyle="sawtooth,pad=0.5",facecolor="beige",alpha=0.3))
plt.axis("off")
plt.show()
```
