## Quiz-7

<span style="color:red">A library contains 4 identical copies of book A, 2 identical copies of book B and 5 identical <br>
copies of book C. These 11 books are arranged on a shelf in the library.<br>
Calculate the number of different arrangements if all the books A are next to each other and <br>
none of the books B are next to each other .complimentary rule 

We are given:
- 4 identical copies of book A
- 2 identical copies of book B
- 5 identical copies of book C
We need to arrange all 11 books such that:
1. All 4 books A are next to each other.
2. The 2 books B are not next to each other.
Step 1: Group the 4 books A together
Since all A books must be next to each other, treat them as one block:
\[
[AAAA], B, B, C, C, C, C, C
\]

We now have 8 objects:
- 1 A-block
- 2 identical B books
- 5 identical C books
Step 2: Arrange the objects without restriction
The number of arrangements is:
$[
\frac{8!}{2!5!}
]
[
=168
]$

Step 3: Subtract arrangements where the two B books are together
Treat the two B books as one block:
\[
[AAAA], [BB], C,C,C,C,C
\]

We now have 7 objects:
- 1 A-block
- 1 B-block
- 5 identical C books
The number of arrangements is:
$[
\frac{7!}{5!}=42
]$

Step 4: Calculate the required arrangements
Subtract the arrangements where the B books are together:
4[
168-42=126
]$

Final answer: 126 arrangements
Key technique: When a condition says two identical objects must NOT be together, calculate:
$[
\text{Total arrangements}-\text{Arrangements where they are together}
]$

## Quiz-8

Find the number of different ways that 5 boys and 6 girls can stand in a row if no boy stands 
next to another boy.

We are given 5 boys and 6 girls, and no two boys can stand next to each other.
Step 1: Arrange the 6 girls first
Since all 6 girls are different, they can be arranged in:
$[
6!=720
]$

ways.
Step 2: Identify the available spaces for the boys
After arranging the girls, there are 7 spaces:
\[
\_ \ G \ \_ \ G \ \_ \ G \ \_ \ G \ \_ \ G \ \_ \ G \ \_
\]

To ensure that no two boys stand next to each other, each boy must occupy a different space.

We have 7 spaces and need to choose 5:
\[
\binom{7}{5}=\frac{7!}{5!2!}=21
\]

Step 3: Arrange the 5 boys
The 5 boys are different, so they can be arranged in:
\[
5!=120
\]

ways.
Step 4: Calculate the total
Using the multiplication principle:
\[
6!\times\binom{7}{5}\times5!
\]

$[
=720\times21\times120
][
=\boxed{1,814,400}
]$

Final answer: 1,814,400 ways
Remember this technique: When no two people of one group can stand together, arrange the other group first, then place the first group into the gaps.

### python solution

```python
math.perm(6,6)*math.perm(7,5)
```

## Quiz-9

Freddie has 6 toy cars and 3 toy buses, all different. He chooses 4 toys to take on holiday with 
him.<br>
Find the number of possible arrangements if there is a car at each end of the line and no buses 
are next to each other.

<hr>
We are given:
- 6 different toy cars (C)
- 3 different toy buses (B)
Freddie chooses 4 toys and arranges them in a line.

Conditions:
1. There must be a car at each end.
2. No two buses can be next to each other.
<hr>
Step 1: Consider the possible combinations
Since there must be a car at each end, we need at least 2 cars.

We consider three cases:
- 2 cars and 2 buses
- 3 cars and 1 bus
- 4 cars and 0 buses
<hr>
Step 2: Calculate each case
Case 1: 2 cars and 2 buses

- Choose and arrange 2 cars at the ends:
$[
{}^6P_2=6\times5=30
]$

- Arrange the 2 different buses in the middle:
\[
2!=2
\]

However, the two buses would be next to each other, which is not allowed.

Therefore:
$[
0
]$
<hr>
Case 2: 3 cars and 1 bus

- Choose and arrange 2 cars at the ends:
$[
{}^6P_2=30
]$

- Choose 1 more car for the middle:
$[
4
]$

- Choose 1 bus:
$[
3
]$

- Arrange the middle two toys:
$[
2!=2
]$

Total:
$[
30\times4\times3\times2=720
]$
<hr>
Case 3: 4 cars and 0 buses
$[
{}^6P_4=6*6*4*3=360
]$
<hr>
- Choose and arrange 2 cars at the ends:
$[
{}^6P_2=30
]$

- Choose and arrange 2 more cars in the middle:
$[
{}^4P_2=4\times3=12
]$
<hr>
Total:
$[
30\times12=360
]$
<hr>
Step 3: Add all cases
$[
0+720+360=1080
]$
Final answer: 1080 arrangements
