```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
n_rows  =  6
n_cols  =  4
data    =  np.zeros((  n_rows,   n_cols  ))
data 

for i in range(n_cols):
    for j in range(n_rows):
        data[j,i]=np.arange(1,n_rows+1)[j]

df=pd.DataFrame(data,index=np.arange(1,7,1),
                columns=["First_Failure","Second_Failure","Third_Failure","Fourth_Failure"])
df.loc["Probaility",:]=5/6
df.loc["Probaility","Fourth_Failure"]=1/6
df["Prob"]=df.product(axis=1)
df=df.round({ "First_Failure" :2,
              "Second_Failure":2,
              "Third_Failure" :2,
              "Fourth_Failure":2,           
              "Prob":5}             )
df
df1=df.iloc[0:6,0:4].map(lambda x: str(int(x)))
df2=pd.DataFrame(df.iloc[6,0:4].map(lambda x:str(x)))
df3=pd.DataFrame( df.loc[:,"Prob"].map( lambda x:str(x) ) )
df2=df2.T
df_all1=pd.concat((df1,df2),axis=0)
df_all2=pd.concat((df_all1,df3),axis=1)
df_all2.loc[:6,"Prob"]=["prob of 1st 6 on the 4th roll a 6-sided die?",
                        "$p(X=x)=q^{X-1} \\cdot p $",
                        "|Geometric Distribution| x = 4",
                        "p = 1 '6' out of six-sided $\\frac{1}{6}$ q=1-p ",
                        "multiple 2dice 6*6=36 matrix 4 rolls- ",
                        "nd1vector 2matrix 3cluster 4groups"]
df_all2
```

```python
data_g = np.random.geometric(p=1/6, size=120000)
plt.figure(figsize=(12,6))
sns.kdeplot(data_g,lw=4,ls="-",color="green")
plt.title("Geometric Distribution",size=30,color="blue")
plt.annotate("Discrete probability distribution  \ndescribes the number of trials \nneeded to achieve the first success in a\nsequence of independent Bernoulli trials, \nwhere each trial has only two outcomes: \nsuccess or failure, \nand the probability of success.\n\n\nformula $p(X=x)=q^{X-1} \\cdot p $",
             size=15,color="indigo",
             bbox=dict(boxstyle="sawtooth,pad=0.5",facecolor="cyan",alpha=0.4),
             xytext=(30,0.05),
             xy=(10,0.033),
             arrowprops=dict(arrowstyle="fancy",facecolor="magenta",connectionstyle="arc3,rad=0.2")
             )
plt.savefig("sdfd.pdf")
plt.show()
```
