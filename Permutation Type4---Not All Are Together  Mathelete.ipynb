## Quiz-1

The 9 letters of the word HAPPINESS are arranged in random order in a line.<br>
Find the probability that the 3 vowels (A, E, I) are not all next to each other. 

We are given the word HAPPINESS:
\[
H,A,P,P,I,N,E,S,S
\]

There are 9 letters, with:
- 2 identical P's
- 2 identical S's
- 3 vowels: A, E, I
We need to find the probability that the 3 vowels are NOT all next to each other.
Step 1: Find the total number of arrangements
Since P appears twice and S appears twice:
$[
\text{Total}=\frac{9!}{2!\times2!}
][
=90720
]$

Step 2: Find the number of arrangements where all 3 vowels are together
Treat the 3 vowels (A, E, I) as one block.

We now have 7 objects:
$[
[AEI],H,P,P,N,S,S
]$

The number of ways to arrange these 7 objects is:
$[
\frac{7!}{2!\times2!}=1260
]$

The 3 vowels can also be arranged within their block in:
$[
3!=6
]$

Therefore:
$[
\text{Arrangements with all vowels together}
=1260\times6
=7560
]$

Step 3: Find the probability that all 3 vowels are together
$[
P(\text{all together})=\frac{7560}{90720}
=\frac{1}{12}
]$

Step 4: Find the probability that they are NOT all together
Using the complementary probability:
$[
P(\text{not all together})=1-P(\text{all together})
]$

$[
=1-\frac{1}{12}
]
[
=\boxed{\frac{11}{12}}
]$

Final answer: $(\frac{11}{12})$

Key technique: When a question asks for the probability that something does NOT happen, it is often easier to calculate the probability that it DOES happen first, then subtract from 1.

### python solution

```python
num = math.perm(9,9) / (math.factorial(2) * math.factorial(2))
num
```

```python
ads =math.perm(7)/(math.factorial(2)*math.factorial(2))
```

```python
btt =math.factorial(3)
```

```python
den = ads*btt
```

```python
np.around((num-den)/num,4)
```

We need to find the conditional probability:
$[
(\text{Rs together}\mid \text{all 3 Es together})
]$

Since we are given that all three Es are together, we only consider arrangements where the Es form one block.
Step 1: Treat the three Es as one block
The word RESERVED contains:
- R appears twice.
- E appears 3 times.
- S, V, D each appear once.
Treat EEE as one block.

We now have 6 objects:
$[\textnormal{
(EEE), R, R, S, V, D}
]$

The total number of arrangements is
$[
\frac{6!}{2!}=360
]$

Step 2: Count arrangements where both Rs are together
Treat the two Rs as one block (RR).

We now have 5 objects:
$[\textnormal{
(EEE), (RR), S, V, D}
]$

The number of arrangements is
$[
5!=120
]$

Step 3: Calculate the probability
$[
P(\text{Rs together}\mid \text{Es together})
=\frac{120}{360}
]$

$[
=\frac{1}{3}
]$

Final answer: $(\boxed{\frac13}$)

Exam tip: 
<br>In conditional probability questions, the word given tells you to restrict your sample space first. 
<br>Here, we treat the three Es as one block before counting anything else.

```python
# Non_distinct items :                             :{'E': 3, 'R': 2}
# Arrangement without counting non_distinct items  :40320
# Non_distinct items factorial:12
# final answer :3360.0

math.perm(5,5)/3360
```
