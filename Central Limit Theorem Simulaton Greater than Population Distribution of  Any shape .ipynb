```python
import numpy as np 
import matplotlib.pyplot as plt
import seaborn as sns
```

# Population distribution is not normal n >30  

```python
np.random.seed(2)
population= np.random.randint(low=100,high=120,size=10000)
plt.figure(figsize=(12,6))
sample_size= 400 # sample size n >30
sample_time= 300
lst=[]
for i in range(1,sample_time,1):
    lst.append(np.mean([np.random.choice(population,sample_size,replace=True)]))
sns.kdeplot(population)
sns.kdeplot(lst)
plt.title("CLT Simulation using Random Int")
plt.annotate("CLT if sample size n>30 no matter what shape\n the population is \nsampling distribution of \nsample means or proprotions will be normal",
             xytext=(96,0.5),color="yellow",size=12,
             bbox=dict(boxstyle="sawtooth,pad=0.5",facecolor="indigo",edgecolor="crimson",alpha=0.9),
             xy=(109,0.06),
             arrowprops=dict(arrowstyle="fancy",facecolor="cyan",edgecolor="indigo",connectionstyle="arc3,rad=0.2") )
plt.xticks(np.arange(95,125,1),rotation=90)
plt.savefig('CTL.pdf')
plt.show()
```

# population is normal n<30

```python
plt.figure(figsize=(12,6))
np.random.seed(3)
x=np.random.randn(10000)
sns.kdeplot(x)

sff=1000
szn=10
smeans=[[],[],[],[]]
for i in range(4):
    for j in range(sff):
        smeans[i].append(np.mean(np.random.choice(x,szn+i*2,replace=False)))
for i in range(4):
    sns.kdeplot(smeans[i],lw=0.8,label="Sample Size={}".format(szn+i*2))
plt.legend()
plt.title("CLT : population normal n<30")
plt.xticks(np.arange(-5,5,1))
plt.plot([-1.2,1.2],[0.2,0.2],ls="--",color="grey")
plt.plot([-0.2,0.25],[1,1],ls="--",color="grey")
plt.annotate("",xytext=(-2,0.75),
             xy=(-1.2,0.2),
             arrowprops=dict(arrowstyle="fancy",facecolor="red",edgecolor="black",connectionstyle="arc3,rad=0.2"))
plt.annotate("",xytext=(-1.1,0.75),
             xy=(-0.2,1),
             arrowprops=dict(arrowstyle="fancy",facecolor="cyan",edgecolor="indigo",connectionstyle="arc3,rad=0.2"))
plt.annotate(text="Sampling distribution of sample Means\nSample std vs population std",xy=(-4,0.75),
             xytext=(-4,0.75),color="indigo",
             bbox=(dict(boxstyle="sawtooth,pad=0.5",facecolor="white",edgecolor="black")))
plt.axvline(0,0,0.95,ls="--",lw=4,color="grey",alpha=0.5)
plt.axvline(0,0,0.95,ls="-.",lw=1,color="blue")
plt.annotate("mean of Population data =\n mean of Sample means",xy=(0,0.6),
             xytext=(2,0.8),color="indigo",
             bbox=dict(boxstyle="sawtooth,pad=0.5",facecolor="cyan",edgecolor="black"),
             arrowprops=dict(arrowstyle="fancy",facecolor="beige",edgecolor="red",connectionstyle="arc3,rad=0.2"))
plt.savefig('sdf.pdf')
plt.show()
```

```python
size1=50000
plt.figure(figsize=(12,6))
data=np.random.randn(size1)
sns.kdeplot(data,label="Population Distribution Normal")
plt.legend()
samplesize=[5,20,40]
samplefreq=10000

means=[[],[],[]]
for i in range(samplefreq):
    for j in range(len(samplesize)):
        means[j].append(np.mean(np.random.choice(data, samplesize[j],replace=False)))
for i in range(len(means)):
    sns.kdeplot(means[i],label="Sampling Distribution n :{}".format(samplesize[i]))
    plt.legend(facecolor="cyan")
plt.savefig("xx.pdf")
```
