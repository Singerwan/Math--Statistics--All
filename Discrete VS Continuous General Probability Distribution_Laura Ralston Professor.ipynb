# 4 rules of probability_Laura_Ralston_Probability 

<ol>
    <li> All probabilities are between 0 and 1 </li>
        <ul>
            <li>Probabilities are never negative</li>
            <li>Probabilities are never greater than 1 or 100%</li>
        </ul>
    <li> The probability of an impossible event is 0 . P(impossible)=0 </li>
        <ul>
            <li>P(roll a 7 on a die)=0 </li>
            <li>P(Thanksgiving is on Sunday)=0  </li>
        </ul>
    <li> The probability of a certain event is 1 . P(certian)=1 </li>
        <ul>
            <li>P(roll a number that's less than 7 on a die)=1 </li>
            <li>P(Easter is on Sunday)=1  </li>
            <li>0---------------------50|50---------------------1<br>
            Impossible----------50|50---------------------Certain</li>            
        </ul>
    <li> The sum of the probabilities for all the possible options is 1 </li>

### exaample 1 shift nurse _Laura_Ralston_Probability 

Does this table represent a probability distribution? <br>
A hospital researcher is interested in the number of times  <br>
the average post-op patient will ring the nurse during a  <br>
12-hour shitf . <br>
For a random sample of 50 patients , the following information <br>
was obatained. <br>
Let X = the number of times a patient rings the nurse during a  <br>
12-hour shift . <br>
The possible outcomes ,x=0,1,2,3,4,5 <br>

```python
df=pd.DataFrame(data = {"x": np.arange(6,dtype="object"),
                        "P(x)":list(map(lambda x: "P{}=".format(str(x)) ,np.arange(6))),
                        "Decimal": [0.08,0.16,0.32,0.28,0.12,0.04],
                        "comments":["Two requirements from 4 Rulse",
                                    "for checking whether a table is a discrete probability Distribution",
                                    "First $\\Sigma=1$ Second :range(0-1) ",
                                    "Codes for checking : first highlight in cyan if its correct",
                                    "second font-color red if its wrong",""]},
                index=map(lambda x: int(x) ,np.arange(1,7)))
df.loc["Total","Decimal"]=df["Decimal"].sum()

def row_form(r):              # reqruiement of total =1
    if r.loc["Decimal"] ==1 :
        return ["background-color:cyan"]*len(r)
    else:
        return [""]*len(r)       

(df.fillna("")).style.apply(row_form,axis=1).\
                map(subset="Decimal",func= lambda x:"color:red" if x<0 or x>1 else"color:white").\
                format("{:.2f}%",subset="Decimal").\
                set_table_styles([{"selector":"th","props":"border:solid cyan 2px"},
                                  {"selector":"td","props":"border:solid yellow 2px"}])
# reqruiement range
```

#### example 2 -attrition rate laura-ralston

A company wants to evaluate its attrition rate , in other words <br>

let P(x) = the probability that a new hire will stay with the company x years. <br>

```python
df=pd.DataFrame(data = {"x": np.arange(7),
                        "P(x)":list(map(lambda x: "P{}=".format(str(x)) ,np.arange(7))),
                        "Decimal": [0.12,0.18,0.3,0.15,0.1,0.1,0.05],
                       },
                index=map(lambda x: int(x) ,np.arange(1,8)))
val_5=df.query("x>=5")['Decimal'].sum()
val_4=df.query("x<4")['Decimal'].sum()
mean =(df['Decimal']*df["x"]).sum()
var  =( ( (df["x"]-mean)**2) *df["Decimal"]).sum()
np.sqrt(var)
mean
df["$\\mu$"]=(df["x"]*df["Decimal"]).sum()
df["x-$\\mu$"]=df["x"]-df["$\\mu$"]
df["x_mu_diff_sqr"]=df["x-$\\mu$"]**2
df["var_x"]=df["x_mu_diff_sqr"]*df["Decimal"]
df["$\\Sigma$"]=df["var_x"].sum()
df["std"]=np.sqrt(df["$\\Sigma$"])
df.round(2)
```
