$[

## Quiz-3

The back row of a cinema has 12 seats, all of which are empty. A group of 8 people, including <br>
Mary and Frances, sit in this row. Find the number of different ways they can sit in these 12 <br>
seats if 
(ii) Mary and Frances do not sit in seats which are next to each other 

### python solution

We can use the complementary method:
$[
\text{Required arrangements}
=
\text{Total arrangements}
-
\text{Arrangements where Mary and Frances sit together}
]$
<hr>
Step 1: Find the total number of arrangements<br>
There are 12 seats and 8 different people.

Choose and arrange 8 people in 12 seats:
$[
{}^{12}P_8=\frac{12!}{4!}
]$
<hr>
Step 2: Find arrangements where Mary and Frances sit next to each other<br>
There are 11 possible pairs of adjacent seats:
$[
(1,2),(2,3),\ldots,(11,12)
]$

Mary and Frances can swap places in 2 ways.

The remaining 6 people can be arranged in the remaining 10 seats:
$[
{}^{10}P_6
]$

Therefore:
$[
11\times2\times{}^{10}P_6
]$
<hr>
Step 3: Subtract
$[
\begin{aligned}
&{}^{12}P_8-11\times2\times{}^{10}P_6\\
&=19\,958\,400-11(2)(151\,200)\\
&=19\,958\,400-3\,326\,400\\
&=\boxed{16\,632\,000}
\end{aligned}
]$
<hr>
Final answer
 16,632,000 ways

Key technique: When a question says two people must NOT sit together,<br> it is often easiest to calculate the total arrangements and subtract the arrangements where they ARE together.

```python
math.perm(12,8)-math.perm(10,6)*11*2
```

## Quiz-4

Find the number of ways all 10 letters of the word COPENHAGEN can be arranged so that  <br>
(ii) the Es are not next to each other. 

### python solution

```python
def wordarrangement(word):
    dic  = sorted(Counter(word).items() , key =lambda item : item[1] , reverse=True)
    non_dis_dic={}
    non_dis_lst=[]
    for k , v in dic :
        if v >1:
            non_dis_dic[k]=v
            non_dis_lst.append(v)
    den=1
    for i in non_dis_lst:
         den=den*math.perm(i,i)

    num =math.perm(len(word),len(word))
    result = num/den
    print("Non_distinct items :                             :{}\nArrangement without counting non_distinct items  :{}\nNon_distinct items factorial:{}\nfinal answer :{}".format(non_dis_dic, num, den , result))
```

```python
word = "COPENHAGEN"
wordarrangement("COPENHAGEN")
```

## Quiz-11

Find the probability that a randomly chosen arrangement of the 9 letters in the word CELESTIAL <br>
does not have the two Es together. 

We use the complementary counting method:
\[
P(\text{Es not together})=1-P(\text{Es together})
\]

Step 1: Count all possible arrangements
The word CELESTIAL has 9 letters:
- E appears twice.
- L appears twice.
- C, S, T, I, A appear once.
Total arrangements:
$[
\frac{9!}{2!\times2!}=90720
]$

Step 2: Count arrangements where the two Es are together
Treat the two Es as one block (EE).

We now have 8 objects:
- EE (one block)
- L, L
- C, S, T, I, A
Number of arrangements:
$[
\frac{8!}{2!}=20160
]$

Step 3: Find the probability that the Es are together
$[
P(\text{Es together})
=\frac{20160}{90720}
=\frac{2}{9}
]$

Step 4: Find the probability that the Es are NOT together
$[
P(\text{Es not together})
=1-\frac{2}{9}
]$

Final answer: $(\frac{7}{9})$

Exam tip: For probability questions involving letters that must not be together, it is usually easier to calculate:
$[
1-\frac{\text{Arrangements together}}{\text{Total arrangements}}
]$

### python solution

```python
# Non_distinct items :                             :{'E': 2, 'L': 2}
# Arrangement without counting non_distinct items  :362880
# Non_distinct items factorial:4
# final answer :90720.0

1-(math.perm(8,8)/math.perm(2,2))/90720
```
