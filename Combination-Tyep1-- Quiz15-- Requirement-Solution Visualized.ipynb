# Mathlete Quiz Combination Type 1 Quiz 15

Q15 (S21/qp53/q6(d)) <br>
Five of the 11 letters in the word REQUIREMENT are selected. <br>
How many possible selections contain at least two Es and at least one R?

```python
math.comb(6,2)+math.comb(6,1)+math.comb(6,1)+1 
# ========================================== 3 es 2 rs
```

```python
word  = "REQUIREMENT"
letter_list=[x for x in  word]
dic   = sorted(Counter(word).items(), key= lambda item : item[1], reverse=True)

uniq_lett=[]
for k,v in dic:
    if v ==1:
        uniq_lett.append(k)

n_unq= len(uniq_lett)
n_Es=dict(dic)["E"]
n_Rs=dict(dic)["R"]
len(word)
```

```python
matrix     =np.zeros((4,6),dtype="object")
cases_list =[   ["E","E","R","1","2","^6C_2"],
                ["E","E","R","R","1","^6C_1"],
                ["E","E","E","R","1","^6C_1"],
                ["E","E","E","R","R","^6C_0"]  ] 
for i in range(matrix.shape[0]):
    matrix[i]=cases_list[i]
matrix
```

```python
plt.figure(figsize=(18,9))
plt.xlim(-1,11)
plt.ylim(0,7)
x1=np.arange(0,11)/2.5
y1=np.ones(11)*5.5

for i in range(len(x1)):
    plt.plot(x1[i],y1[i],marker="${}$".format(word[i]),ms=40,mew=1.5,mec="green")   
    plt.plot(x1[i],y1[i]+0.6,marker="o".format(i),ms=30,mec="indigo")  
    plt.plot(x1[i],y1[i]+0.6,marker="${}$".format(i),ms=20,mew=1,mec="blue")   
    if i in [0,5]:
        plt.plot(x1[i],y1[i]-0.6,marker="*".format(i),ms=30,mec="black",mew=3)  
    elif i in [1,6,8]:
        plt.plot(x1[i],y1[i]-0.6,marker="^".format(i),ms=20,mec="white",mew=3)  

rect=plt.Rectangle((-0.5,4.5),5,2.1,facecolor="cyan")
plt.gca().add_patch(rect)

x2=np.arange(0,4.9,0.8)
y2=np.arange(0.7,0.7+0.95*4,0.95)

for i in range(matrix.shape[0]):
    for j in range(matrix.shape[1]):
        if j <5:
            plt.plot(x2[j],y2[i],marker="p",ms=65,mew=2,mec="blue")
            plt.plot(x2[j],y2[i]-0.43,marker="_",ms=65,mew=5)
            plt.plot(x2[j],y2[i],marker="${}$".format(matrix[i,j]),ms=30,mec="black")            
        else:
            plt.plot(x2[j]+1.2,y2[i]-0.35,marker="1",ms=20,mew=2)
            plt.plot(x2[j]+0.5,y2[i],marker="s",ms=60,mew=2,mec="indigo")
            plt.plot(x2[j]+0.5,y2[i],marker="${}$".format(matrix[i,j]),ms=50,mec="black")

plt.annotate("Five of the 11 letters in the word \nREQUIREMENT are selected. \nHow many possible selections contain \nat least two Es and at least one R?",
             xytext=(6,1),size=20,color="blue",
             bbox=dict(boxstyle="sawtooth,pad=0.5",facecolor="yellow",edgecolor="magenta"),
             xy=(8.32,4.65),
             arrowprops=dict(arrowstyle="simple",facecolor="green",connectionstyle="arc3,rad=0.00001"))

plt.annotate("11 letters in total 3Es + 2Rs =5 n=6(#unique letters)\nCase1 $^6C_0$=1\nCase1 $^6C_1$=6\nCase1 $^6C_2$=6\nCase1 $^6C_2$=15 $\\Sigma$=1+6+6+15=28" 
             ,xytext=(6,4.8),size=15,color="magenta",
             bbox=dict(boxstyle="square,pad=0.5",facecolor="beige",edgecolor="blue"),
             xy=(4.5,5.45),
             arrowprops=dict(arrowstyle="simple",facecolor="brown",connectionstyle="arc3,rad=0.0001"))
plt.axis("off")
plt.savefig("sdfds.pdf")
plt.show()
matrix.shape
```
