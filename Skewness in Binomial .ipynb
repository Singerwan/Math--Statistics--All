```python
colorlist = []  #color list
for i in range(10000):
    colorlist.append('#%06X' % np.random.randint(0, 0xFFFFFF))  # colors list
```

```python
data_sk=np.zeros((20,3))
df_sk_S=pd.DataFrame(data_sk,index=np.arange(20))
df_sk_S.rename(columns={0:"Symmetric",1:"Left_skewed",2:"Right_skewed"})

arr1= np.array((list(sorted(np.linspace(0.01,0.2,6),reverse=True)))[1:])
arr2= np.array(list(np.linspace(0.01,0.2,6)))
arra= np.append(arr2,arr1)
arra

df_sk_S.loc[5:15,"Symmetric"]       =arra
df_sk_S.loc[:10,"Left_skewed"]      =[0.        , 0.08333333, 0.16666667, 0.25,0.17 , 0.136, 0.102, 0.068, 0.05,0.034, 0.]
df_sk_S.loc[9:,"Right_skewed"]      =[0.        , 0.034,0.05,0.068,  0.102,0.136 , 0.17 ,0.25,0.16666667, 0.08333333,0.02]

# df_sk_S.loc[10:,"Right_skewed"]     =arra

fig,axes=plt.subplots(2,3,figsize=(18,6))

for k in range(2):
    for j in range(3):
        for i in range(len(df_sk_S['Symmetric'])):
            axes[0][0].scatter(x=np.arange(0,20)[i],y=df_sk_S['Symmetric'][i],marker="o",s=30)
            axes[0][1].scatter(x=np.arange(0,20)[i],y=df_sk_S['Left_skewed'][i],marker="p",s=30)
            axes[0][2].scatter(x=np.arange(0,20)[i],y=df_sk_S['Right_skewed'][i],marker="*",s=30)
            axes[k][j].set_xlim(-1,21)
            axes[k][j].set_ylim(-0.02,0.3)
            axes[k][j].set_xticks(np.arange(0,21))    
for i in range(3):
    for j in range(2):
        axes[1][i].axis("off")

annotations1=["Symmetric Tossing a coin \n20 times:\n where p=p(heads)",
             "Skewed_to_right \nTossing a die 20 times \nwhere p=p(getting a 6)",
             "Skewed_to_left- \nTossing a die 20 times \nwhere p=p(not getting a 6)",]
annotations2=["\\mu=10 \\quad ",
             "\\mu=3\\frac{1}{3} ",
             "\\mu=16\\frac{2}{3} ",]
annotations3=["\\sigma=2.236",
             " \\sigma=1\\frac{2}{3} \\text{why-left hump is smaller}",
             "\\sigma=1\\frac{2}{3} \\text{why-right hump is samller}",]

for i in range(3):
    for j in range(2):
        axes[1][i].annotate("{}".format(annotations1[i]), xy =(0,0.16),size=15,color="black",bbox=dict(boxstyle="sawtooth,pad=0.3",facecolor=colorlist[i+7]))
        axes[1][i].annotate("${}$".format(annotations2[i]), xy =(0,0.1),size=13,color="black",bbox=dict(boxstyle="sawtooth,pad=0.3",facecolor=colorlist[i+7]))
        axes[1][i].annotate("${}$".format(annotations3[i]), xy =(0,0.03),size=13,color="black",bbox=dict(boxstyle="sawtooth,pad=0.3",facecolor=colorlist[i+7]))
df_sk_S['Symmetric'].sum()

plt.show()
```
