## Quiz_4.1

- The probability that a student at a large music college plays in the band is 0.6.
- For a student who plays in the band, the probability that she also signs in the choir is 0.3.
- For a student who doesn't play in the band, the probability that she signs in the choir is x.
- The probability that a randomly chosen student from the college dosn't sign in the choir is 0.58.
  - Find the value of x
  - Two Students from the college are chosen at random
    - find the probability that both students play in the band<mark> and </mark>both sing in the choir 

- how many ways can a student attend a choir - either in the band or not in the band 
- B  0.6
  - C  0.3
  - NC 0.7     ---------- ( 0.6 * 0.7 )        play in the band and sing in the choir   
- NB 0.4                                        
   - C  x
   - NC 1-x    ---------- (0.4 * ( 1-x ) )

Find the probability that one student plays in the band AND sings in the choir.<br>
$[
P(B\cap C)=P(B)\times P(C\mid B)
][
=0.6\times0.3
][
=0.18
]$

Step 3: Find the probability for both students.
Assuming the two students are chosen independently:
$[
P(\text{both play in band and sing in choir})
][
=0.18\times0.18
][
=0.0324
]$

Final answer
$[
\boxed{0.0324}
]$

Or as a percentage:
$[
\boxed{3.24\%}
]$

### python solution 

```python
import sympy as sympy
```

```python
x=sympy.symbols("x")

eq=sympy.Eq(0.58,(0.6*0.7)+0.4*(1-x))
```

```python
eq
```

```python
sympy.solve(eq,x)[0]
```

```python
data =[[0.3*0.6,0.7*0.6],
      [0.6*0.4,0.4*0.4]]
df=pd.DataFrame(data=data,
                index =["B","NB"],
                columns=['C','NC']) 
df['total']=df.sum(axis=1)
df.loc['total']=df.sum(axis=0)
df=df.map(lambda x : "{:.2f}%".format(x*100))
df.style.set_table_styles([{"selector":"th","props":"color:red;border:double white"},
                           {"selector":"td.col0.row0","props":"background-color:cyan;color:red"}]).\
         set_properties(**{"border":"dotted yellow 3px"})
```

## Quiz_4.2

- On each day that Alexa goes to work.The probability that she travels by bus, by train or by car are
- 0.4, 0.35, 0.25 respectively.
- When she travels by bus , the probability that she arrives late is 0.55.
- When she travels by train, the probability that she arrives late is 0.7.
- When she travels by car , the probability that she arrives late is x.
- On a randomly chosen day , when Alexa goes to work the probability that she doesn't arrive late is 0.48.
  - Find the value of x


- B  0.4
  - Late  0.55
  - NLate 0.45     ---------0.4*0.45
- T 0.35                                    
  - Late  0.7
  - NLate 0.3     ---------0.35*0.3
- C  0.25
  - Late  x
  - NLate 1-x     ---------0.25*(1-x)
- total Nlate= 0.48

### python solution 

```python
import sympy as sympy
x1  = sympy.symbols("x1")
eq1 = sympy.Eq(0.48,0.45*0.4 + 0.35*0.3 +0.25*(1-x1))
sympy.solve(eq1,x1)
```
