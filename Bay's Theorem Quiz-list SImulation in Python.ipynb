## formula overview 

<div style="background-color:indigo;color:lightteal">
<hr>&emsp;&emsp;&rlarr; Rules of Elimination : If B1,B2,B3.. are mutually exclusive events of which 1 must occur <hr> &emsp;&emsp;&rlarr;then $ \ P(A) \ =\ \Sigma^{n}_{i=1} \ [P(B_{i}) \cdot P(A|B_{i})] $ &emsp; $\Sigma$ Summation | or |<br> &emsp;&emsp;&rlarr;P(A) = (0.6)(0.95) + (0.3)(0.8) + (0.1)(0.65)= 0.875 <br><hr>
 &emsp;&emsp;&rlarr; Bay's Theorem P(B3|A)&emsp;&rarr;&emsp;P(A $\cap$ B3) =P(A) $\cdot$ P(B3|A) <br>&emsp;&emsp;&rarr;P(B3|A)=  $\frac{P(A \  \cap \   B3)}{P(A) }$<hr>
 &emsp;&emsp;&rlarr;$ P(A   \ \cap \     B3) = P(B3) \cdot P(A|B3) = P(A) \cdot P(B3|A) $&rarr; 4 times 3 = 3 times 4<hr>
<div style="font-size:20px;background-color:brown;font-weight:bold"> &emsp;&emsp;$ P(B3|A)=\frac{P(B3\cdot)P(A|B3)} {\Sigma^{n}_{i=1}P(Bi) \cdot P(A|Bi)}=\frac{1}{all}$</div><hr></div>

```python
# 1 cars
prob_cars  = np.array( [ 0.6,  0.3 , 0.1  ] )
prob_pasc  = np.array( [ 0.95, 0.8 , 0.65 ] )
subbranch  = prob_cars * prob_pasc 
total      = subbranch.sum()
proportion = subbranch/total

df         = pd.DataFrame(  data = {"probability_cars"                :prob_cars , 
                                    "probbality_pass_rate_given_cars" :prob_pasc ,
                                    "Sunbranch_probability"           :subbranch ,
                                    "Sum"                             :total ,
                                    "Proportion"                      :proportion ,},
                            index = ['Plant_B1','Plant_B2','Plant_B3'])
print("Question 1: If i buy a car , what's the probability that it will be problem free?                        ans :{:.2f}%\nQuestion 2: if somebody buys a car , what's the probability that plant B3 produced the problem free car? ans :{:.2f}%"\
    .format(df.iloc[0,-2]*100,df.iloc[2,-1]*100))
df.round(3)
```

```python
# 1 Three vendoer supply materials
com_prob  = np.array([ 0.3,  0.4 , 0.3  ])  # first step calculate the probability of individual event occuring| supply of material share| - total to 1 
def_rate  = np.array([ 0.02, 0.03, 0.05 ])  # second calculate the probaility given that event has occur | company abc defct |
prob_d_N  = (com_prob * def_rate)           # third calculate each indivdual subracnh ;s probability of defect rate given that the product is produced by company  
prob_d_n  = (com_prob * def_rate).sum()     # fourth add up subbranch's probability 
prob_com_d= prob_d_N/prob_d_n               # fifth find out the proportion of A def over all ABC def

df=pd.DataFrame(data={ "Company_Probability" : com_prob ,
                       "Defect_Rate" : def_rate ,
                       "Probability of defect given its from company n" : prob_d_N ,
                       "Sum" : prob_d_n,
                       "Proportion" : prob_com_d },
                index=["A","B","C"] )
print(df.iloc[:,-1])
df.round(2)
```

```python
# 2 three candicates for professor 
prob_post   =np.array( [4,2,3]) /np.array( [4,2,3]).sum() # first to calculate the candicate's probability - and the sum should be 1 100% pie chart 
prob_nsub   =np.array( [0.35, 0.52 , 0.80] )              # second probability of each - subbranch - given that candidates become professor - 
prob_ps_ns  =prob_post*prob_nsub                          # third to calculate each branch's probability - using multiplication along the subbranch
prob_p_ns_t =prob_ps_ns.sum()                             # fourth add up all subbranches's probability based on given condition met
prof_ns_prob=prob_ps_ns/prob_p_ns_t                       # fifth to calculate the proportion in respect to the total -conditions met - 

df=pd.DataFrame(data={ "Candidate_probability_of_prof":prob_post,
                       "prof_probability_of_subj":prob_nsub,
                       "prof_nsubjt_probability":prob_ps_ns,
                       "Sum":prob_p_ns_t,
                       "Prob_cand_prof_nsub":prof_ns_prob} ,
                index=["X","Y","Z"])
df.round(3)
```

```python
# 3 three bags 
prob_3bags  = np.array( [1/3,1/3,1/3] ) 
data_colr   = np.zeros( (3,2) )

# prob_balls  = np.array([])
data_colr   = np.array([   [2,1],
                           [3,2],
                           [1,3]] )
data_colr=data_colr.reshape(3,-1)
r_to_choose = 1
r_all_to_ch = 2
total_n_1bg = 6

comb_ret    = []

for i in range(data_colr.shape[0]):
        comb_ret.append(np.around(  ( math.comb(data_colr[i][0],r_to_choose  )
                                     * math.comb(data_colr[i][1],r_to_choose) )
                                          / math.comb(total_n_1bg,r_all_to_ch )        ,3)  )
comb_ret=np.array(comb_ret)

prob_3bags_ball=  prob_3bags * comb_ret
sum_p          =  prob_3bags_ball.sum()

sub_prob       =  prob_3bags_ball /sum_p

df             =  pd.DataFrame(data= {  "probability_3bags"         :prob_3bags,
                                        "probability_balls_in3bags" :comb_ret ,
                                        "probability_balls_givenbag":prob_3bags_ball ,
                                        "Sum "                      :sum_p ,
                                        "proportion "               :sub_prob ,
                                   },
                                index= ["Bag1","Bag2","Bag3"])
print("Probability of coming from the first bag is  :{:.2f}%".format(df.iloc[0,-1]*100))
df.round(2)
```

```python
# 4 bridge collapse
default_prob  = np.array( [ 0.1  ,0.9  ] )
colp_giv_pro  = np.array( [ 0.95 ,0.45 ] )

subbranch_pro = default_prob * colp_giv_pro
total         =  subbranch_pro.sum()
prop          =  subbranch_pro/total 

df= pd.DataFrame( data = {"Faulty_probability"        :default_prob  ,
                          "collapse_given_faulty_prob":colp_giv_pro  ,
                          "subbranch"                 :subbranch_pro  ,
                          "sum"                       :total ,
                          "proportion"                :prop  ,},
                  index =['DesigningError','InferiorMaterial'])
print("DesigningError probability is  {:.2f}%".format(df.iloc[0,-1]*100))
df
```

```python
# 5 coin false true
n_trials = 4
true_prob= 1/2
fals_prob= 1/1
# construction 
prob_coins = np.array ([  3/4,                  1/4])
prob_heads = np.array ([  true_prob**n_trials ,
                          fals_prob**n_trials   ]   )
subbranch  = prob_coins * prob_heads
total      = subbranch.sum()
proportion = subbranch/total

df         = pd.DataFrame(data = {"probability_coins"     :prob_coins ,
                                  "probability_heads"     :prob_heads ,
                                  "subbranch_probability" :subbranch ,
                                  "Sum"                   :total ,
                                  "proportion"            :proportion ,} ,
                          index=['True_Coin','False_Coin']
                                   )
print("Probability of coming from the false coin is :{:.2f}%".format(df.iloc[1,-1]*100))
df.round(2)
```

```python
# 6 bag balls 

prob_bag = np.array( [ 1/2, 1/2 ]) 
prob_bal = np.array( [ math.comb(3,1)/math.comb(9,1), 
                       math.comb(6,1)/math.comb(9,1) ])
subbrach = prob_bag * prob_bal
total    = subbrach.sum()
prop     = subbrach/total

df       = pd.DataFrame(data  = {   "probability_bag"  :prob_bag , 
                                    "probability_ball" :prob_bal , 
                                    "subbranch_prob"   :subbrach , 
                                    "Sum"              :total , 
                                    "proportion"       :prop , },
                        index = ['Bag1','Bag1'])

print("The proability that the ball drawn is of white color is  :{:.2f}%".format(df.iloc[0,-2]*100))
df.round(2)
```
