```python
# The Organic Chemistry Tutor 1
e=np.e
t=np.arange(20)
lambd =1/5
x=np.linspace(0,20,1000)
y=lambd * e **(-(lambd * x))
fig,ax=plt.subplots(1,2,figsize=(12,4))



ax[0].plot(x,y,ls="-",lw=3,color="black")

x_fil_exp =np.linspace(0,3,100)
y_fil_exp =stats.expon.pdf(x_fil_exp,scale=5)
ax[0].fill_between(x_fil_exp, y_fil_exp, color="cyan",alpha=0.3)

x_fil_exp1 =np.linspace(4,7,100)
y_fil_exp1 =stats.expon.pdf(x_fil_exp1,scale=5)
ax[0].fill_between(x_fil_exp1, y_fil_exp1, color="red",alpha=0.3)

x_fil_exp2 =np.linspace(10,20,100)
y_fil_exp2 =stats.expon.pdf(x_fil_exp2,scale=5)
ax[0].fill_between(x_fil_exp2, y_fil_exp2, color="indigo",alpha=0.3)
ax[0].set_xticks(np.arange(19))

ax[0].axvline(3,0.05,0.5,color="red",ls="--",lw=3)
ax[0].axvline(4,0.05,0.4,color="indigo",ls="--",lw=3)
ax[0].axvline(7,0.05,0.27,color="indigo",ls="--",lw=3)
ax[0].axvline(10,0.05,0.15,color="brown",ls="--",lw=3)
ax[0].set_yticks(np.arange(0,0.21,0.05))


x= np.linspace(0,20,1000)
l= 1/5
t= np.arange(1,15)
e= np.e
y= l*e**(-l*x)

ind=np.arange(1,8,0.5)
for i in range(len(ind)):
    ax[1].plot(x,l*e**(-l*x)*ind[i],color=colorlist[i])
    ax[1].annotate("Scaling factor is : {}".format(ind[i]),xy=(12,1.4-(i*0.07)),color=colorlist[i])

plt.title("PDF $f(x)=\\lambda \\cdot e^{-\\lambda \\cdot x}$ CDF $1-e^{-\\lambda \\cdot x}$",x=0,size=25,color="red")
plt.show()
```

```python
from scipy.stats import expon
import matplotlib.pyplot as plt
import seaborn as sns

#When average time between 2 messages is 50 seconds
data1 = expon.rvs(scale=50, size=10000)

#When average time between 2 messages is 60 seconds
data2 = expon.rvs(scale=60, size=10000)

#When average time between 2 messages is 80 seconds
data3 = expon.rvs(scale=80, size=10000)

#Plot sample data
sns.kdeplot(x=data1, fill=True, label='1/lambda=50')
sns.kdeplot(x=data2, fill=True, label='1/lambda=60')
sns.kdeplot(x=data3, fill=True, label='1/lambda=80')
plt.xlabel('Units of time between successive events')
plt.ylabel('Probability')
plt.title('Exponential Distribution')
plt.legend()
plt.xlim(0, 200)
plt.show();
```
