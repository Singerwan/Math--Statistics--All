Q3 (W15/qp62/q4(i))<br>
A group of 8 friends travels to the airport in two taxis, P and Q. Each taxi can take 4 passengers. <br>
i. The 8 friends divide themselves into two groups of 4,
<br>one group for taxi P and one group 
for taxi Q, with Jon and Sarah travelling in the same taxi. <br>
Find the number of different ways in which this can be done. 

```python
from imojify import imojify
from matplotlib import pyplot as plt 
from matplotlib.offsetbox import OffsetImage,AnnotationBbox

dfig,ax=plt.subplots(1,1,figsize=(12,6))
ax.plot(0,0,"")
ax.set_xlim(-1,9)
ax.set_ylim(0,6)

emjisss = ["👱🏻‍♀️👩🏻‍🦰👩🏻👧👨🏻‍💼👱🏻‍♀️👩🏻‍🦰👩","🚕","🚗","👱🏻‍♀️👩🏻‍🦰👩🏻👧👨🏻‍💼👱🏻‍♀️","👩🏻‍🦰👩"]
x=[7,0,3,1.5,1.5]
y=[0.5,3.5,3.5,1.3,4.87]
for i in range(len(emjisss)):
    img = plt.imread(imojify.get_img_path(emjisss[i]))
    if i <1:
        ax.add_artist( AnnotationBbox(
                                       OffsetImage(
                                                    img, zoom=0.06),
                       (x[i], y[i]),  frameon=False, pad=0.5))
    elif i ==1 or i<3:
        ax.add_artist( AnnotationBbox(
                               OffsetImage(
                                            img, zoom=0.15),
               (x[i], y[i]),  frameon=True, pad=0.5))
    else  :
        ax.add_artist( AnnotationBbox(
                               OffsetImage(
                                            img, zoom=0.06),
               (x[i], y[i]),  frameon=False, pad=0.5))
        

ax.annotate("",xytext=(1.6,4),
            xy=(2.36,2.75), arrowprops=dict(arrowstyle="fancy",facecolor="cyan"))
ax.annotate("$^2C_1$",xytext=(1.25,4),size=20,color="red",
            bbox=dict(boxstyle="sawtooth,pad=0.3",facecolor="yellow"),
            xy=(0.65,2.75), arrowprops=dict(arrowstyle="fancy",facecolor="magenta"))
ax.annotate("$^6C_2$",xytext=(1.25,1.9),size=20,color="red",
            bbox=dict(boxstyle="sawtooth,pad=0.3",facecolor="cyan"),
            xy=(0.65,2.75), arrowprops=dict(arrowstyle="fancy",facecolor="brown"))
ax.annotate("$^6C_4$",xytext=(1.8,2.3),color="white",size=11,
            bbox=dict(boxstyle="circle",facecolor="green",edgecolor="magenta",linewidth=2),
            xy=(2.36,2.75), arrowprops=dict(arrowstyle="fancy",facecolor="blue"))
ax.annotate("A group of 8 friends travels to the airport in 2 taxis(PQ).\nEach taxi can take 4 passengers.\nThe 8 friends divide themselves into two groups of 4,\none group for taxi P and one group for taxi Q, \nwith Jon and Sarah travelling in the same taxi.\nFind the number of different ways \nin which this can be done."
        ,xy=(5,4.5),color="black",size=8,weight="bold",
           bbox=dict(boxstyle="square,pad=0.5",facecolor="yellow",edgecolor="red",linewidth=3))
ax.annotate("logic\nFirst 2(SARA,JOHN)choose 1 of the two different taxis.\n\n1out of the remaning 6  choose 2 to join sara and john\n2out of the remaning 6 choose 4 for the next taxi"
        ,xy=(5,3),color="maroon",size=8,weight="bold",bbox=dict(boxstyle="LArrow,pad=0.5",facecolor="cyan",edgecolor="blue",linewidth=3))
plt.show()
```
