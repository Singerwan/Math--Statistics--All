## Quiz-4

Find how many different numbers can be made by arranging all nine digits of the 
number 223 677 888 if <br>
(i) there are no restrictions, <br>
<span style="background-color:cyan;color:indigo">(ii) the number made is an even number.</span>

### python solution

```python
math.perm(9,9)/(math.factorial(2)*math.factorial(2)*math.factorial(3))
```

<hr>
<div style="font-size:15px">We need to find how many <span style="background-color:indigo;color:white">different even numbers </span>can be made using all nine digits:
[2,2,3,6,7,7,8,8,8] <br>
An even number must end in an even digit. Therefore, the last digit must be 2, 6 or 8.<br>

We consider three cases.<br>
<hr>
Case 1: Last digit is 2 The remaining digits are:|\[2,3,6,7,7,8,8,8\]
There are 8 digits, with two 7s and three 8s repeated.<br>
$[\frac{8!}{2!\times3!}=3360]$
<hr>
Case 2: Last digit is 6 The remaining digits are: \[2,2,3,7,7,8,8,8\]
There are two 2s, two 7s and three 8s repeated.<br>
$[\frac{8!}{2!\times2!\times3!}=1680]$
<hr>
Case 3: Last digit is 8<br>
The remaining digits are:<br>
\[2,2,3,6,7,7,8,8\]
There are two 2s, two 7s and two 8s repeated.<br>
\[\frac{8!}{2!\times2!\times2!}=5040\]
<hr>
Step 4: Add all three cases<br>
\[3360+1680+5040\]
\[=\boxed{10080}\]

Final answer: 10,080 numbers<br>
Key idea: For an even number, consider every possible even final digit separately, then add the results.<br>
<hr>

```python
eight=(math.perm(8,8)/(math.perm(2,2)*math.perm(2,2)*math.perm(2,2))) # 5040
two  =(math.perm(8,8)/(math.perm(2,2)*math.perm(2,2)*math.perm(3,3))) # 1680
six  =(math.perm(8,8)/(math.perm(2,2)*math.perm(1,1)*math.perm(3,3))) # 3360

eight+two+six
```
