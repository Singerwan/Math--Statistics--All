For B(100, 1/5) n=100 p=1/5 find 
<hr>

- mean var std .
- what n value keeping p the same would allow the<span style="color:cyan"> standard deviation to be less than </span> <span style="color:yellow"> 1% of n .</span>
  -  <span style="color:yellow"> 1 percent of n = n devided by 100  </span>
  -  $\textcolor{yellow}{\frac{n}{100}}$
  -  $\textcolor{cyan}{\text{standard deviation} \quad \sigma = \sqrt{n\cdot p\cdot q}}$
  -  <span style="color:brown;background-color:white">construct the formula  |can't use n =100 std=4 because n is supposed to be unknown|</span>
  -   $\textcolor{red}{\sqrt{n \cdot p \cdot q} \lt \frac{n}{100}} $
  -   rewrite to get rid of square root solving  plugin p=$\frac{1}{5}$ q=1-p
  -   $\textcolor{lighblue}{npq<(\frac{n}{100}})^2 \rightarrow n\cdot\frac{1}{5}\cdot\frac{4}{5}<\frac{n^2}{10000}$
  -  <span style="font-size:20px"> $ n \cdot \frac{4}{\cancel{25}^{1}} <\frac{n^2}{\cancel{10000}^{400}}$</span>
  -   $4\cdot 400 \cdot n< n^2  \rightarrow 1600n < n^2$
  -   0 < n(n-1600)) ans n>1600 |can not be less than 0 - as there's no negative |
- keeping n=100 , what would p need to be increased or decreased to so that standard deviation was less than 1 ?  Are both options possible?
  - $\sqrt{npq}<1 \rightarrow \sqrt{np(1-p)}<1$
  - np(1-p)<1 plug in n =100
  - $100p-100p^2 <1$  &rar; $100p-100p^2 -1 <0$
  - change signs to get a parabola quadratic formula back by changing the signs to flip less than to greater than
  - $100p^2 -100p +1 >0$
  - two options are both possible - a quardracic formula 

## Solution

```python
import sympy
p=sympy.symbols("p")
ex=sympy.Eq(0,100*p**2-100*p+1)
sympy.plot(100*p**2-100*p+1)
result_list=sympy.solve(ex,p)
```
