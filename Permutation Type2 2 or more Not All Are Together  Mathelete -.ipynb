<div style="color:red">6.(S13/qp62/q6(ii))
[3]<br>
A town council plans to plant 12 trees along the centre of a main road. The council buys the <br>
trees from a garden centre which has 4 different hibiscus trees, 9 different jacaranda trees and 2  <br>
different oleander trees for sale.  <br>
The council buys 4 hibiscus trees, 6 jacaranda trees and 2 oleander trees.  <br>
(ii) How many different arrangements of these 12 trees can be made if the hibiscus trees have to  <br>
be next to each other, the jacaranda trees have to be next to each other and the oleander trees  <br>
have to be next to each other?  <br>

We have 12 different trees:
- 4 different hibiscus trees (H)
- 6 different jacaranda trees (J)
- 2 different oleander trees (O)
We want trees of the same type to be next to each other.
<hr>
Step 1: Treat each type of tree as one block
Since each type must stay together, we have 3 blocks:
$[\textnormal{
(HHHH),\quad(JJJJJJ),\quad(OO)
}]$

These 3 blocks can be arranged in:
$[
3!=6
]$
ways.
<hr>
Step 2: Arrange the trees within each block
All trees are different, so:
- Hibiscus trees: \(4!\)
- Jacaranda trees: \(6!\)
- Oleander trees: \(2!\)
<hr>
Step 3: Multiply everything together
$[
3!\times4!\times6!\times2!
]
[
=6\times24\times720\times2
]
[=\boxed{207360}
]$
<br>
Final answer: 207,360
Exam tip: When different categories of objects must each stay together, 
<br>treat each category as a single block. Then multiply the arrangements of the blocks by the arrangements within each block.

### python solution

```python
math.perm(6,6)*math.perm(4,4)*math.perm(2,2)*math.perm(3,3)
```

$[

$[

\frac{4!}{3!}=4

The consonants and vowels occupy separate fixed positions, so:
$[
12\times4=\boxed{48}
]$

Final answer: 48

## Quiz-17

<div style="color:red">17.(S17/qp62/q6(ii)) <br>
[3]
A library contains 4 identical copies of book A, 2 identical copies of book B and 5 identical <br>
copies of book C. These 11 books are arranged on a shelf in the library.<br>
Calculate the number of different arrangements if all the books A are next to each other and <br>
none of the books B are next to each other. 


We have:
- 4 identical books A
- 2 identical books B
- 5 identical books C
<hr>
  <br>
Conditions:<br>
1. All 4 books A must be together.<br><br>
2. The 2 books B must not be next to each other.
<hr>
Step 1: Treat the 4 A books as one block<br>
Since all the A books must be together, treat AAAA as one object.<br>

We now have:
$[\textnormal{
(AAAA), B, B, C,C,C,C,C}
]$

That gives us 8 objects, with 2 identical Bs and 5 identical Cs.
<hr>
Step 2: Find the total number of arrangements
Ignoring the restriction on the Bs for now:
$[
\frac{8!}{2!5!}
]$

$[
=\frac{40320}{2\times120}=168
]$
<hr>
Step 3: Subtract arrangements where the Bs are together
Treat the two Bs as one block (BB).

We now have:
$[\textnormal{
(AAAA), (BB), C,C,C,C,C}
]$

This gives 7 objects, with 5 identical Cs.

The number of arrangements is:
$[
\frac{7!}{5!}
]$

$[
=\frac{5040}{120}=42
]$
<hr>
Step 4: Calculate the required arrangements
$[
168-42=\boxed{126}
]$
<hr>
Final answer: 126
<hr>
Exam tip: 
<hr>
When a question says two types of objects must not be together,<br
                                                                   > first count all arrangements, then subtract the arrangements where they are together.

### python solution

```python
total =math.perm(8,8)/ (math.perm(2,2)*math.perm(5,5))
total
```

```python
btog= math.perm(7,7)/math.perm(5,5)
```

```python
total-btog
```
