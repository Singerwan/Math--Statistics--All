## Visualized

```python
data1 = np.zeros((6,6),dtype="object")
data2 = np.zeros((6,6),dtype="object")
die1 = np.arange(1,7)
die2 = np.arange(1,7)
ydec = [7,7,7,7,7,7]
plt.figure(figsize=(12,9))

plt.ylim(-2,9)
plt.yticks(np.arange(-2,9,0.4),rotation=0)
plt.xlim(-2,7)
plt.xticks(np.arange(-2,7,0.2),rotation=90)

for i in range(len(die1)):
    plt.plot(i,ydec[i]+0.5,marker="s",ms=70,alpha=1,mew=3,mec="blue")
    for j in range(len(die2)):
        data1[i,j]="{},{}".format(die1[i],die2[j])
        data2[i,j]=int(die1[i] + die2[j])
        plt.plot(i,j,marker="s",ms=70,alpha=1,mec="black",mew=3,mfc="white")
        plt.plot(i,j,marker="s",ms=70,alpha=0.1,mec="blue",mew=4)

        plt.plot(i,j,marker="${}.{}={}$".format(die1[i],die2[j],die1[i]+die2[j]),ms=50)
        plt.plot(i,j-0.5,marker="$Die1,Die2\\Sigma$",ms=40,alpha=1,mec="black",mew=0.3)        

plt.axis("off")
plt.plot([0,0.7,1.3,2,2,2, 2.7,2.7,3.3,3.3,   3.7,3.7,4.3,4.3, 4  , 4.7,5,5.3,4.7,5,5.3
         ],
         [7.5,8,7,7,7.5,8, 7,8,7,8 ,   7,8,7,8,7.5,  7,7,7,8,8,8,
           ],marker="o",ls="",color="black",ms=10,mec="black")
plt.annotate("$e^n$ One Die(1,2,3,4,5,6) $6^1$=6  Two Dice $6^2$=36 |Coin(H-T)| $2^n$",
             xy=(-0.5,6.1),size=15,color="indigo",
             bbox=dict(boxstyle="sawtooth,pad=0.3",facecolor="magenta"))
plt.show()
```

## Python solutions

### 2 Dice -Doublet

```python
data = np.zeros((6,6))
die  = np.arange(1,7)

n1 =  3
n2 =  11
count =0
for i in range(len(die)):
    for j in range(len(die)):
        data[i][j]=int(die[i]+die[j])
        if die[i]==die[j]:
            count=count+1
print("doublet number is :{}".format(count))
list(data.reshape(36,-1)).count(n1)
list(data.reshape(36,-1)).count(n2)  
```

### 2 Dice Sum 

```python
# roll 2 dice
data = np.zeros((6,6,6))
die  = np.arange(1,7)

n=5
for i in range(len(die)):
    for j in range(len(die)):
        for k in range(len(die)):
            data[i][j][k]=die[i]+die[j]+die[k]
print("3 Dice - probability of getting {}{}/216".format(n,list(data.reshape(6*6*6,-1)).count(n)))
```

### 6 Dice Sum

```python
# roll 6 dice 
data = np.zeros((6,6,6,6,6,6))
die  = np.arange(1,7)

for i in range(len(die)):
    for j in range(len(die)):
        for k in range(len(die)):
             for l in range(len(die)):
                for m in range(len(die)):
                    for n in range(len(die)):
                        data[i][j][k][l][m][n]=die[i]+die[j]+die[k]+die[l]+die[m]+die[n]
                        
list(data.reshape(6*6*6*6*6*6,-1)).count(10)
```
