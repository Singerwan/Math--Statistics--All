# Mathlete Combinations Type4 Quiz12

Q12 (W15/qp63/q5(b)) --------
<br>
There are 
<hr>
    7 Chinese, 
<br>6 European and 
<br>4 American students at an international conference. 
<hr>
    Four of the students are to be chosen to take part in a television broadcast. 
<br>Find the number of  different ways the students can be chosen if 
<br> at least one Chinese and at least one European  student are included. 

```python
## Python Solution 
```

```python
math.comb(7,1)*math.comb(6,1)*math.comb(4,2) +\
math.comb(7,1)*math.comb(6,2)*math.comb(4,1) +\
math.comb(7,1)*math.comb(6,3)*math.comb(4,0) +\
math.comb(7,2)*math.comb(6,1)*math.comb(4,1) +\
math.comb(7,2)*math.comb(6,2)*math.comb(4,0) +\
math.comb(7,3)*math.comb(6,1)*math.comb(4,0) 
df =pd.DataFrame(data= {"Ch_n":np.ones(6)*7,"Ch_r":[1,1,1,2,2,3],
                        "Eu_n":np.ones(6)*6,"Eu_r":[1,2,3,1,2,1],
                        "Am_n":np.ones(6)*4,"Am_r":[2,1,0,1,0,0],})
df=df.map(lambda x :int(x))
quiz12val=df.values
result= []
latex = []
for i in range(6):
    result.append(   math.comb( list( df['Ch_n'] )[ i ], list( df['Ch_r'] )[ i ] )*\
                        math.comb( list( df['Eu_n'] )[ i ], list( df['Eu_r'] )[ i ])*\
                        math.comb( list( df['Am_n'] )[ i ], list( df['Am_r'] )[ i ]))
    latex.append( "$^{}C_{} \\times ^{}C_{} \\times ^{}C_{}  $".\
                  format(
                      int( list( df['Ch_n'] )[ i ] ),
                      int( list( df['Ch_r'] )[ i ] ),
                      int( list( df['Eu_n'] )[ i ] ),
                      int( list( df['Eu_r'] )[ i ] ),
                      int( list( df['Am_n'] )[ i ] ),
                      int( list( df['Am_r'] )[ i ] )    )   )
new_col={"result":result,
         "formula":latex,
         "$\\Sigma$":np.array(result).sum()}
df=df.assign(**new_col)
df.style.set_properties(subset="formula", **{"backkground-color":"indigo","color":"cyan","border":"double red 2px","padding":"8px"}).\
         set_properties(subset="result", **{"background-color":"blue","color":"yellow","border":"double white 2px","padding":"8px"}).\
         set_properties(subset="$\\Sigma$", **{"background-color":"red","color":"yellow","border":"double white 2px","padding":"8px"}).\
         highlight_max(subset="Am_r",color="red").highlight_min(subset="Am_r",color="green").\
         highlight_max(subset="Ch_r",color="crimson").highlight_min(subset="Ch_r",color="blue").\
         bar(subset="Eu_r",cmap="flag")
```
