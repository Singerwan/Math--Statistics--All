# Library

```python
import numpy as np
import pandas as pd
from itertools import permutations 
import math
```

# Jeep

## Overview

```python
lstE=  ["J","E", "E", "P"]
lE12=  ["J","E1.","E2.","P"]
```

$\frac{4!}{(4-4)! \cdot 2! }=\frac{4!}{1 \cdot 2! }=\frac{24}{2}=12$

```python
dist=math.perm(len(lstE)) # numerator  4 chooses 4 in permuation arrangement 
dup_E=math.factorial(2)   # denominator 2 factorial - duplicated value E occurs 2 times 
dist/dup_E                # final answer 
```

## Duplicated

```python
permE= permutations(lstE)
peE12= permutations(lE12)

lstE=[]
for i in permE:
    lstE.append(i)
lE12=[]
for j in peE12:
    lE12.append(j)

dfE=pd.DataFrame(lstE,index=np.arange(1,25),columns=['E_001','E_002','E_003','E_004'])
dfE['ConcE00']=dfE['E_001']+dfE['E_002']+dfE['E_003']+dfE['E_004']
dfE
dfE12=pd.DataFrame(lE12,index=np.arange(1,25),columns=['E12_1','E12_2','E12_3','E12_4'])
dfE12['ConcE12']=dfE12['E12_1']+dfE12['E12_2']+dfE12['E12_3']+dfE12['E12_4']
dfE12

df_all_jeep_je1e2p=pd.concat((dfE,dfE12),axis=1)
df_all_jeep_je1e2p['ID']=np.arange(1,25)

df_all_jeep_je1e2p
odd_index=[i for i in np.arange(1,25) if i%2!=0]
eve_index=[i for i in np.arange(1,25) if i%2==0]

def odd_ind(x):
    if x in odd_index:
        return "background-color:cyan;color:indigo"
    elif x in eve_index:
        return "background-color:red;color:green"

df_all_jeep_je1e2p.style.map(odd_ind)
```

## Unique

```python
df_jeep_je1e2p_unq =df_all_jeep_je1e2p[df_all_jeep_je1e2p['ConcE00'].duplicated()]
df_jeep_je1e2p_unq.index=np.arange(1,13)
df_jeep_je1e2p_unq
```
